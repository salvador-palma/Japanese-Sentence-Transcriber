# ETL9B Repacker and Organizer

## (Local) Repacking ETL9B

```python
import pandas as pd
import os
import tqdm



def CreateParquet(id):
    def ReadImgBytes(index):
        filename = f"Data/ETL9B_{id}_unpack/{index:05d}.png"
        if os.path.exists(filename):
            with open(filename, 'rb') as f:
                return f.read()
        return None

    df = pd.read_csv(f'Data/ETL9B_{id}_unpack/meta.csv')

    df['image_data'] = [ReadImgBytes(i) for i in tqdm.tqdm(range(len(df)))]

    df.to_parquet(f'Data/dataset{id}.parquet', index=False)

#Arrange each folder that stemed from ETL9B in a parquet file
for id in range(1, 6):
    CreateParquet(id)
    print(f"File 'dataset{id}.parquet' created successfully.")
```

```python
import glob
import pandas as pd

#Join all parquet files into a big one

keep = ['char', 'image_data'] 
parquets = glob.glob('Data/dataset*.parquet')

dfs = []

for file in parquets:
    print(f"Reading {file}...")
    df = pd.read_parquet(file, columns=keep)
    dfs.append(df)

allDf = pd.concat(dfs, ignore_index=True)

allDf.to_parquet('Data/final_processed_dataset.parquet', index=False)
```
